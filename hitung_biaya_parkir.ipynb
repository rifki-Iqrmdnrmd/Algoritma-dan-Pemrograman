```python
# Program Hitung Biaya Parkir - Python

# Parameter NIM (26552011042)
V = 2

# 1. Input Data
jeis_kendaraan = input("Masukkan jenis kendaraan (M/K/B): ").upper()
# Menggunakan float dan mengganti koma dengan titik agar input desimal aman
durasi_input = input("Masukkan durasi parkir (jam): ").replace(',', '.')
durasi = float(durasi_input)
is_member_input = input("Apakah Anda member? (y/n): ").lower()
is_member = is_member_input == 'y'

# 2. Penentuan Tarif Dasar (SWITCH-CASE / match-case)
match jeis_kendaraan:
    case 'M':
        tarif_dasar = 5000 + (V * 100)   # Rp 5.200
    case 'K':
        tarif_dasar = 2000 + (V * 100)   # Rp 2.200
    case 'B':
        tarif_dasar = 10000 + (V * 100)  # Rp 10.200
    case _:
        print("Jenis Kendaraan Tidak Valid")
        exit()

# 3. Percabangan Durasi & Denda (Nested IF / IF-ELSE IF)
if durasi <= 2:
    total_tagihan = durasi * tarif_dasar
elif durasi <= 8:
    biaya_awal = 2 * tarif_dasar
    biaya_lanjut = (durasi - 2) * (tarif_dasar * 0.9)  # Diskon 10%
    total_tagihan = biaya_awal + biaya_lanjut
else:  # durasi > 8 jam
    biaya_awal = 2 * tarif_dasar
    biaya_menengah = 6 * (tarif_dasar * 0.9)           # Diskon 10% untuk 6 jam berikutnya
    biaya_lanjut = (durasi - 8) * (tarif_dasar * 0.8)  # Diskon 20% untuk sisa jam
    denda = V * 1000                                   # Rp 2.000
    total_tagihan = biaya_awal + biaya_menengah + biaya_lanjut + denda

# 4. Percabangan Member (Single IF)
if is_member:
    total_akhir = total_tagihan - (total_tagihan * 0.15)  # Diskon 15%
else:
    total_akhir = total_tagihan

# Output
print(f"Total Biaya Parkir Anda: Rp {total_akhir:,.2f}")
```
